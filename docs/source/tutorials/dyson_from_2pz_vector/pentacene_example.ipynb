# Pentacene STM reference calculation

This is the existing three-state pentacene example, using supplied HOMO/LUMO site coefficients and model carbon 2pz orbitals. It provides a regression baseline for library development. The [example guide](README.md) records the assumptions, units, and command-line instructions. This is not a multireference or experimental benchmark.

````python
%matplotlib inline
from IPython.display import display
import matplotlib.pyplot as plt
from pentacene_calculation import example_parameters, run_calculation, plot_results
````

## Calculation parameters

The source plane and tip z are absolute coordinates in the supplied geometry frame. Gaussian HWHM is a phenomenological broadening, not a temperature. The returned dictionary can be edited before running.

````python
parameters = example_parameters()
parameters
````

## Run the shared calculation

The command-line runner calls the same function. Currents are retained in amperes; figures display nanoamperes. Rates include the Clebsch-Gordan multiplet factors.

````python
result = run_calculation(parameters)
assert result["report"]["passed"], result["report"]
figures = plot_results(result)
for figure in figures.values():
    plt.close(figure)
````

## I(V) at the selected tip position

Both lead currents use the tip-to-substrate sign convention and should agree. The requested lateral position is snapped to the nearest grid point, as in the original example.

````python
print("Requested x/y (Å):", parameters["point_xy_angstrom"])
print("Sampled x/y (Å):", result["arrays"]["point_sampled_xy_angstrom"])
display(figures["iv"])
````

## Constant-height current map

The figure displays the current magnitude at -0.6 V; the signed numerical map is preserved. The cyan cross marks the sampled I(V) position.

````python
display(figures["current_map"])
````

## State populations

GS, D0-, and D0+ denote neutral, anion, and cation multiplets. This example has no excited neutral states or additional relaxation channels.

````python
display(figures["populations"])
````

## Numerical checks

Checks cover finite values, normalized nonnegative populations, the scaled stationary-equation residual, equal lead currents, and a separately evaluated zero-bias point. These establish internal consistency; the tests also check the analytic three-state solution and the original notebook snapshot.

````python
result["report"]
````
