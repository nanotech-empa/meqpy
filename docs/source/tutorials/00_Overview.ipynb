# Meqpy Tutorial — Overview

``meqpy`` is a Python package for rate-equation (master-equation) simulations of transport experiments, with a particular focus on scanning tunneling microscopy (STM) and spectroscopy (STS). This tutorial walks through the package chapter by chapter, from the basic Markov-chain formalism up to spatially resolved charging rates derived from Dyson orbitals.

Repository: [github.com/nanotech-empa/meqpy](https://github.com/nanotech-empa/meqpy)

## How to use this tutorial

The chapters build on one another and are best read in order, with one exception: Chapter 4, *BandSystem and BandTransitions*, is not a prerequisite for Chapter 5, *Spatial Resolution*. Each notebook has a navigation bar at the top and bottom linking to the previous/next chapter and back to this index.

Chapter 5 ("Spatial Resolution") is split across three notebooks (**5a**, **5b**, **5c**) because of its length — 5a covers the ``Cube``/``Transition`` base classes, 5b covers the ``Molecule`` class, and 5c ties everything together in worked examples. They should be read as a single chapter, in that order.

## Chapters

1. **[Rate Equation: Basics](01_Basics.ipynb)** — The master-equation formalism, ``meqpy.solve_equilibrium``, and how anchor states and ``fuzz`` are used to obtain a unique solution.
2. **[System and States](02_System_and_States.ipynb)** — Defining ``State`` and ``System`` objects, and calculating charging rates (normalized charging transitions, coupling strength, spin selection rule).
3. **[Simulating Experiments](03_Simulating_Experiments.ipynb)** — Putting it together to simulate I(V), I(z), STML, and photocurrent experiments.
4. **[BandSystem and BandTransitions](04_BandSystem_BandTransitions.ipynb)** — Extending the model to systems with continuous electronic bands (e.g. bulk or 2D materials) using ``BandSystem`` and ``BandTransition``.
5. **Spatial Resolution** *(three-part chapter — read in order)*
    - **[5a. Cubes and Transitions](05a_Cubes_and_Transitions.ipynb)** — Loading volumetric orbital data (``Cube``, ``Cube_2pz``) and the abstract ``Transition``/``Dyson`` classes for spatially resolved charging.
    - **[5b. Molecule](05b_Molecule.ipynb)** — The ``Molecule`` class: registering Dyson transitions and calculating spatially resolved charging rates.
    - **[5c. Molecule Examples](05c_Molecule_Examples.ipynb)** — Worked examples: constant-height STS maps and I(V) point spectroscopy on a molecule.

## Prerequisites

- A working installation of ``meqpy`` (see below)
    - including dependencies: ``numpy``, ``ase``, ``scipy``
- ``matplotlib`` for displaying results

No prior familiarity with rate-equation/master-equation methods is assumed; the formalism is introduced from scratch in chapter 1.

## Installation

Install meqpy from github using:
```bash
pip install git+https://github.com/nanotech-empa/meqpy
```

or by cloning the repo and installing in developer mode:
```bash
git clone https://github.com/nanotech-empa/meqpy
cd meqpy
pip install -e ./
```
